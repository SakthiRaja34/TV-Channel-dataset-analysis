```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
import matplotlib.pyplot as plt

# Monthly sales data
sales = [100, 120, 140, 160, 180, 200,
         110, 130, 150, 170, 190, 210]

# Create SARIMA model
model = SARIMAX(
    sales,
    order=(1,1,1),
    seasonal_order=(1,1,1,12)
)

# Fit model
model_fit = model.fit()

# Forecast next 3 values
forecast = model_fit.forecast(steps=3)

print("Forecast:", forecast)
plt.figure(figsize=(10,5))

plt.plot(
    range(1,13),sales,marker="o",label="Original Sales"
)

plt.plot(
    range(13,16),forecast,marker="o",linestyle="--",label="Forecast"
)

plt.title("SARIMA Sales Forecast")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.show()
```

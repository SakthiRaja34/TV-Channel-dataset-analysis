```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

df = pd.read_csv("poly.csv")

x = df[["Experience"]]#feature
y = df["Salary"]

poly = PolynomialFeatures(degree=2)

x_poly = poly.fit_transform(x)

model = LinearRegression()

model.fit(x_poly, y)

new = [[9]]

new_poly = poly.transform(new)

prediction = model.predict(new_poly)

print("Predicted Salary:", round(prediction[0]))

plt.scatter(x,y)
plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

data={
 "roomsq":[10,20,30,40,50,60,70,80],
 "price":[107000,250000,350000,430000,450300,520400,673000,763500]
}

hell=pd.DataFrame(data)

x=hell[["roomsq"]]
y=hell["price"]

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)

m=LinearRegression()
m.fit(x_train,y_train)

new_area=[[1]]
pred=m.predict(new_area)
print("predicted price:",round(pred[0]))

plt.scatter(data["roomsq"],data["price"])
plt.show()
```

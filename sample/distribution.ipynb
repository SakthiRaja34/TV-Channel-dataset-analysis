```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(1, 10, 1000)
print(data)
plt.hist(data, bins=10)
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.binomial(n=10, p=0.5, size=1000)
print(data)

plt.hist(data, bins=11)
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Binomial distribution
data = np.random.binomial(10, 0.5, 1000)

# Create histogram
counts, bins, patches = plt.hist(data, bins=np.arange(-0.5, 11.5, 1))

# Show frequency on top of each bar
for count, patch in zip(counts, patches):
    if count > 0:
        plt.text(
            patch.get_x() + patch.get_width() / 2,
            count,
            str(int(count)),
            ha='center',
            va='bottom'
        )

plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# lambda = average number of events
calls = np.random.poisson(8, 1000)

plt.hist(calls, bins=15)
plt.title("Poisson Distribution - Customer Calls")
plt.xlabel("Calls per Hour")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Apple stock prices
data = {
    "Day": [1, 2, 3, 4, 5],
    "Price": [180, 182, 181, 185, 183]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate 3-day Rolling Average
df["Rolling Average"] = df["Price"].rolling(window=3).mean()

# Print DataFrame
print(df)

# Plot
plt.figure(figsize=(8,5))

# Original Stock Price
plt.plot(df["Day"], df["Price"],
         marker="o",
         label="Stock Price")

# Rolling Average
plt.plot(df["Day"], df["Rolling Average"],
         marker="o",
         linewidth=3,
         label="3-Day Rolling Average")

plt.title("Apple Stock Price with Rolling Average")
plt.xlabel("Day")
plt.ylabel("Stock Price ($)")
plt.xticks(df["Day"])
plt.legend()
plt.grid(True)

plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

# Sample Time Series Data (Sales)
sales = [100, 102, 101, 99, 120,125,134,152,100,102,100,136,150]

# Perform ADF Test
result = adfuller(sales)

# Display Results
print("ADF Statistic :", result[0])
print("p-value       :", result[1])

# Interpretation
if result[1] < 0.05:
    print("Result : Data is Stationary")
else:
    print("Result : Data is Non-Stationary")
```

```python
from statsmodels.tsa.arima.model import ARIMA

sales = [100,120,130,140,150,160,170]

model = ARIMA(sales, order=(1,1,1))

model_fit = model.fit()

forecast = model_fit.forecast(steps=3)

print("Forecast:", forecast)
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

sales = [100,120,140,160,180,200,
         110,130,150,170,190,210]

model = SARIMAX(
    sales,
    order=(1,1,1),
    seasonal_order=(1,1,1,6)
)

model_fit = model.fit()

forecast = model_fit.forecast(steps=5)

print("Forecast:", forecast)
```

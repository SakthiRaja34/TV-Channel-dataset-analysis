```python
import pandas as pd
df=pd.read_csv("messy_data.csv")
print(df)
```

```python
#Data Type Checking
df.dtypes
```

Missing-Value Analysis 

```python
# Count the total missing values
print(df.isnull().sum())

#Check whether missing values exist
print("\nmissing values exist: ",df.isnull().any())


#only show the missing value
missing = df.isnull().sum()
missing[missing > 0]
```

Duplicate Analysis 

```python
#Check total duplicate rows
print(df.duplicated().sum())

#Display duplicate rows
print(df[df.duplicated()])

#Remove duplicate rows
print( df.drop_duplicates())
```

```python
#Display duplicate rows
print(df[df.duplicated()])
```

Unique-Value Analysis

```python
#Find unique values in a column
print(df['Product_Category'].unique())

#Count unique values
print(df['Product_Category'].nunique())

#Find unique values for all columns
print(df.nunique()) #This shows how many unique values each column contains.

#Display frequency of each unique value
print(df['Product_Category'].value_counts())
```

Invalid-Value Detection

```python
#Age
print(df[(df['Age'] < 18) | (df['Age'] > 100)])
```

```python
#Income
df[df['Income'] < 0]
```

```python
#Quantity
df[df['Quantity'] <= 0]

#Sales
df[df['Sales'] < 0]

#Discount
df[(df['Discount'] < 0) | (df['Discount'] > 100)]

#Profit
df[df['Profit'] < 0]

#Product Category
df[~df['Product_Category'].isin(
    ['Electronics', 'Clothing', 'Furniture', 'Grocery']
)]
```

Groupby Analysis

```python
#Product Category — Total Sales
group_analysis = df.groupby("Product_Category")["Sales"].sum()
print(group_analysis)
```

```python
#Product Category — Average Sales
group_analysis = df.groupby("Product_Category")["Sales"].mean()
print(group_analysis)
```

```python
#Product Category — Multiple Analysis
group_analysis = df.groupby("Product_Category").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Quantity": "sum"
})

print(group_analysis)
```

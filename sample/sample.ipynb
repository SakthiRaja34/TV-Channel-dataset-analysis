```python
#central limit theorem
import numpy as np
import matplotlib.pyplot as plt

# 1. Create population
population = np.random.exponential(
    scale=50,
    size=10000
)

# 2. Create sample means
sample_means = []

for i in range(1000):

    sample = np.random.choice(
        population,
        size=30
    )

    sample_mean = np.mean(sample)

    sample_means.append(sample_mean)

# 3. Plot population
plt.hist(population, bins=50)

plt.title("Population Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()

# 4. Plot sampling distribution
plt.hist(sample_means, bins=30)

plt.title("Sampling Distribution of Sample Means")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.show()
```

```python
#Confidence Interval

import numpy as np
from scipy import stats

data = [
    500, 600, 750, 800, 650,
    900, 700, 550, 850, 720,
    680, 770, 620, 810, 730
]

# Sample size
n = len(data)

# Sample mean
mean = np.mean(data)

# Sample standard deviation
std = np.std(data, ddof=1)

# Standard error
se = std / np.sqrt(n)

# 95% Confidence Interval
ci = stats.t.interval(
    confidence=0.,
    df=n - 1,
    loc=mean,
    scale=se
)
print(n)
print("\n Sample Mean:", mean)
print("\n Standard Deviation:", std)
print("\n Standard Error:", se)
print("\n 95% Confidence Interval:", ci)
```

```python
import numpy as np
from scipy import stats

# Sample data
data = [32, 31, 35, 29, 34, 33, 30, 36, 31, 32]

# Hypothesized population mean
mu = 30

# Significance level
alpha = 0.05

# Perform One-Sample t-Test
t_statistic, p_value = stats.ttest_1samp(data, mu)

print("t-statistic:", t_statistic)
print("p-value:", p_value)

# Decision
if p_value < alpha:
    print("Null hypothesis H0")
else:
    print("alternative hypothesis H0")
```

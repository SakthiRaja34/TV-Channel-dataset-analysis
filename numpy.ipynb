```python
import pandas as pd
import matplotlib.pyplot as plt

# Apple stock prices
data = {
    "Day": [1, 2, 3, 4, 5],
    "Price": [180, 182, 181, 185, 183]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate 3-day Rolling Average
df["Rolling Average"] = df["Price"].rolling(window=3).mean()

# Print DataFrame
print(df)

# Plot
plt.figure(figsize=(8,5))

# Original Stock Price
plt.plot(df["Day"], df["Price"],
         marker="o",
         label="Stock Price")

# Rolling Average
plt.plot(df["Day"], df["Rolling Average"],
         marker="o",
         linewidth=3,
         label="3-Day Rolling Average")

plt.title("Apple Stock Price with Rolling Average")
plt.xlabel("Day")
plt.ylabel("Stock Price ($)")
plt.xticks(df["Day"])
plt.legend()
plt.grid(True)

plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

sale=[90,92,95,90,91,92,93,92,91,95]

result=adfuller(sale)

print("ADF statistic : ",result[0])
# print(f"p-value : {result[1]:.10f}")
print("p-value : ",result[1])

if result[1] < 0.05 :
 print("The result is Stationary")

else:
 print("non-stationary")
```

```python
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA

a=[100,120,130,140,150,160,170]

model=ARIMA(a,order=(1,1,1))

model_a=model.fit()

forecast=model_a.forecast(steps=3)

print("Forecast : ",forecast)

plt.figure(figsize=(10,7))

plt.plot(range(1,8),a ,marker="o",linewidth="2",label="Original Data")
plt.plot(range(8,11),forecast,marker="o",linestyle="--",linewidth="2",color="red",label="forecast")
plt.title("ARIMA")
plt.xlabel("time")
plt.ylabel("sales")
plt.legend()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

data = {
    "Hours": [2, 3, 4, 5, 6],
    "Attendance": [60, 70, 75, 85, 90],
    "Marks": [50, 55, 65, 75, 85]
}

df = pd.DataFrame(data)

# Input
x = df[["Hours", "Attendance"]]

# Output
y = df["Marks"]

# Train-Test Split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)

# Create model
model = LinearRegression()

# fit model
model.fit(x_train, y_train)

# Predict new data
new=[[7,96]]
pred = model.predict(new)

print("Predicted Marks:", pred[0])
```

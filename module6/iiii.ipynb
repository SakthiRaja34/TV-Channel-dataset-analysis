```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier

# Dataset
data = {
    "Attendance": ["High", "High", "Medium", "Medium", "Low", "Low"],
    "Study_Hours": ["High", "Low", "High", "Low", "High", "Low"],
    "Result": ["Pass", "Pass", "Pass", "Fail", "Fail", "Fail"]
}

df = pd.DataFrame(data)

# Label Encoding
attendance_encoder = LabelEncoder()
study_encoder = LabelEncoder()
result_encoder = LabelEncoder()

df["Attendance"] = attendance_encoder.fit_transform(df["Attendance"])
df["Study_Hours"] = study_encoder.fit_transform(df["Study_Hours"])
df["Result"] = result_encoder.fit_transform(df["Result"])

# Features and Target
X = df[["Attendance", "Study_Hours"]]
y = df["Result"]

# Train Decision Tree Model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# Predict
new_data = [[0, 0]]  # Attendance = High, Study_Hours = High
prediction = model.predict(new_data)

# Display Result
print("Prediction:", result_encoder.inverse_transform(prediction)[0])
```

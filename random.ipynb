```python
from sklearn.ensemble import RandomForestClassifier
import pandas as pd


df=pd.read_csv("ftrandom.csv")

x=df[["StudyHours","Attendance"]]
y=df["Result"]

model=RandomForestClassifier(n_estimators=10)
model.fit(x,y)

new_data=[[8,75]]
pred=model.predict(new_data)
print(pred)
```

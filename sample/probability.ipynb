```python
import pandas as pd

data = {
    "Student": [1, 2, 3, 4, 5],
    "Study_Hours": [2, 5, 3, 4, 1],
    "Pass": [0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

# P(A): Studied 5 hours
P_A = len(df[df["Study_Hours"] == 5]) / len(df)

# P(B|A): Passed given that student studied 5 hours
P_B_given_A = df[df["Study_Hours"] == 5]["Pass"].mean()

# Multiplication Rule
P_A_and_B = P_A * P_B_given_A

print("P(A) =", P_A)
print("P(B|A) =", P_B_given_A)
print("P(A AND B) =", P_A_and_B)
print("Probability =", P_A_and_B * 100, "%")
```

```python
import pandas as pd

data = {
    "Student": [1, 2, 3, 4, 5],
    "Study_Hours": [2, 5, 3, 4, 1],
    "Pass": [0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

# P(A) → Studied 5 hours
P_A = len(df[df["Study_Hours"] == 5]) / len(df)

# P(B) → Passed
P_B = len(df[df["Pass"] == 1]) / len(df)

# P(A AND B) → Studied 5 hours AND Passed
P_A_and_B = len(
    df[(df["Study_Hours"] == 5) & (df["Pass"] == 1)]
) / len(df)

# Addition Rule
P_A_or_B = P_A + P_B - P_A_and_B

print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A AND B) =", P_A_and_B)
print("P(A OR B) =", P_A_or_B)
print("Probability =", P_A_or_B * 100, "%")
```

```python
import pandas as pd
data={

"Marks":[23,56,47,38,59,60,46,35,58,60,98,56,25,63]
}

df=pd.DataFrame(data)
mean=df["Marks"].mean()
median=df["Marks"].median()
mode=df["Marks"].mode()
print(df)
print("Mean :",mean)
print("Median :",median)
print("mode :", mode)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Product weights
weights = [480, 490, 495, 500, 505, 510, 515, 500, 498, 502,
           495, 508, 512, 499, 501, 505, 507, 493, 496, 504]

# Mean and Standard Deviation
mean = np.mean(weights)
std = np.std(weights)

print("Mean:", mean)
print("Standard Deviation:", std)

# Histogram
plt.hist(weights, bins=8, edgecolor="black")
plt.xlabel("Product Weight (grams)")
plt.ylabel("Number of Products")
plt.title("Product Weight Distribution")
plt.show()
```

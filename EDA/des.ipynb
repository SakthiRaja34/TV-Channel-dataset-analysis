Skewness

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {"customer_income": [5000, 8000, 12000, 15000, 16000, 18000, 20000, 60000]}

df = pd.DataFrame(data)

# Calculate skewness
skewness = df["customer_income"].skew()

print("Skewness:", skewness)

# Check skewness
if skewness > 0:
    print("Data is Right Skewed")

elif skewness < 0:
    print("Data is Left Skewed")

else:
    print("Data is Symmetrical")


# Calculate Mean, Median and Mode
mean = df["customer_income"].mean()
median = df["customer_income"].median()

print("Mean:", mean)
print("Median:", median)

# Plot
sns.kdeplot(df["customer_income"], fill=False)

plt.axvline(mean, linestyle="--", label="Mean",color="red")
plt.axvline(median, linestyle="--", label="Median",color="black")
plt.title("Customer Income Distribution")
plt.xlabel("Customer Income")
plt.ylabel("Density")

plt.legend()
```

```python
import pandas as pd

data = {
    "Values": [30, 40, 65, 72, 80]
}

df = pd.DataFrame(data)

variance = df["Values"].var(ddof=0)
STD=df["Values"].std(ddof=0)

print("Variance:", variance)
print("STD: ",STD)
```

```python
import pandas as pd

data={
    "salary":[10000,20000,25000,25000,32000,32000,45000,55000]
}

df=pd.DataFrame(data)
print("mean: ",df["salary"].mean())
print("Median: ",df["salary"].median())
print("Mode: ",df["salary"].mode())
print("Range_value: ",df["salary"].max()-df["salary"].min())
```

```python
from scipy.stats import binom
import matplotlib.pyplot as plt

# Parameters
n = 10
p = 0.5

# X values
x = range(n + 1)

# Calculate probabilities
y = binom.pmf(x, n, p)

# Create bar chart
plt.bar(x, y)

plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

plt.show()
```

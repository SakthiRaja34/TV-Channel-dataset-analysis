```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(10, 20, 1000)

plt.hist(data, bins=20)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

dice = np.random.randint(1, 7, 10)

plt.hist(dice, bins=np.arange(1, 8) - 0.5, rwidth=0.8)
plt.xlabel("Dice Value")
plt.ylabel("Frequency")
plt.title("Dice Roll - Uniform Distribution")
# plt.xticks(range(1, 7))
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

dice = np.random.randint(1, 7, 1000)

plt.hist(dice, bins=np.arange(0.5, 7.5, 1), rwidth=0.7)

plt.xticks([1, 2, 3, 4, 5, 6])
plt.xlabel("Dice Value")
plt.ylabel("Number of Rolls")
plt.title("Dice Roll - Uniform Distribution")

plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Age": ["20", "21", "22"],
    "Salary": ["25000", "30000", "35000"]
})

df["Age"] = pd.to_numeric(df["Age"])
df["Salary"] = pd.to_numeric(df["Salary"])

print(df)
print(df.dtypes)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Date": ["2026-09-01", "2026-09-02", "2026-09-03"]
})

df["Date"] = pd.to_datetime(df["Date"])

print(df)
print(df.dtypes)
```

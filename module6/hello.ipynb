```python
import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

st.set_page_config(page_title="Student Marks Analysis", layout="wide")

st.title("📚 Student Marks Analysis System")

st.header("Enter Student Details")

name = st.text_input("Student Name")
tamil = st.number_input("Tamil", 0, 100, 0)
english = st.number_input("English", 0, 100, 0)
maths = st.number_input("Maths", 0, 100, 0)
science = st.number_input("Science", 0, 100, 0)
social = st.number_input("Social", 0, 100, 0)

if "students" not in st.session_state:
    st.session_state.students = []

if st.button("Add Student"):

    total = tamil + english + maths + science + social
    average = total / 5

    if average >= 90:
        grade = "A+"
    elif average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    else:
        grade = "F"

    st.session_state.students.append({
        "Name": name,
        "Tamil": tamil,
        "English": english,
        "Maths": maths,
        "Science": science,
        "Social": social,
        "Total": total,
        "Average": round(average, 2),
        "Grade": grade
    })

if len(st.session_state.students) > 0:

    df = pd.DataFrame(st.session_state.students)

    st.header("Student Details")
    st.dataframe(df)

    st.header("Overall Statistics")

    subjects = ["Tamil", "English", "Maths", "Science", "Social"]

    col1, col2, col3, col4 = st.columns(4)

    col1.metric("Highest Mark", int(np.max(df[subjects].values)))
    col2.metric("Lowest Mark", int(np.min(df[subjects].values)))
    col3.metric("Average Mark", round(np.mean(df[subjects].values), 2))
    col4.metric("Students", len(df))

    topper = df.loc[df["Total"].idxmax()]
    st.success(f"🏆 Topper : {topper['Name']} ({topper['Total']} Marks)")

    st.header("Subject Average")

    subject_avg = df[subjects].mean()

    fig, ax = plt.subplots()
    ax.bar(subject_avg.index, subject_avg.values)
    ax.set_ylabel("Average Marks")
    st.pyplot(fig)

    st.header("Student Total Marks")

    fig2, ax2 = plt.subplots()
    ax2.plot(df["Name"], df["Total"], marker="o")
    ax2.set_ylabel("Total")
    st.pyplot(fig2)

    st.header("Subject Contribution")

    fig3, ax3 = plt.subplots()
    ax3.pie(subject_avg.values,
            labels=subject_avg.index,
            autopct="%1.1f%%",
            startangle=90)
    st.pyplot(fig3)
```
